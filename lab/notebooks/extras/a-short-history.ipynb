# A short history of machine learning

> A brief timeline of the ideas and hardware behind today’s methods.

Machine learning draws on statistics, computer science, neuroscience, and advances in computing. This timeline highlights a few developments behind methods covered in the course. It is context, not a list of dates to memorize.

## 1943–1958: early neural models

In 1943, Warren McCulloch and Walter Pitts described a mathematical model of a neuron. It had inputs, weights, a threshold, and an output, but the weights were set by hand.

In 1949, psychologist Donald Hebb proposed that connections between neurons could strengthen through repeated co-activation. His idea helped inspire later learning rules.

In 1958, Frank Rosenblatt published work on the perceptron, a trainable system for classifying patterns. The perceptron learned by adjusting its weights after errors. The original description included a physical implementation.

The perceptron update is closely related to the gradient updates in this course: an error determines how the weights change, and the input scales that change.

## 1960s–1980s: limits and backpropagation

In 1969, Marvin Minsky and Seymour Papert analyzed limitations of single-layer perceptrons, including their inability to represent XOR. Their mathematical result was about that model class. Interest and funding for neural-network research declined for several reasons, including limited computing power and unrealistic expectations.

Researchers had considered multilayer networks, but efficient training methods were still developing. Paul Werbos described backpropagation in a 1974 thesis. David Rumelhart, Geoffrey Hinton, and Ronald Williams helped bring the method wider attention with a 1986 paper showing how multilayer networks could learn internal representations.

In 1989, Yann LeCun and colleagues demonstrated convolutional networks for handwritten digit recognition. Related systems were later used for practical document-reading tasks.

## 1990s–2000s: other methods lead

Training deep networks remained difficult. Gradients could become too small in early layers, and data and computing resources were limited.

Other approaches performed well on the datasets of the time. Support vector machines offered strong results and useful theory. Random forests and gradient boosting became effective methods for tabular data. Long short-term memory networks, introduced in 1997, helped recurrent networks learn longer sequence relationships.

These methods were important advances, not detours. Several remain useful today, especially for small and medium-sized tabular datasets.

## 2006–2012: larger datasets and faster hardware

In 2006, work on layer-wise pretraining helped renew interest in deep networks. Researchers also began using graphics processing units for neural-network training.

ImageNet provided a large labelled image collection and a shared benchmark. In 2012, AlexNet combined convolutional networks, GPU training, and large-scale labelled data to achieve a major improvement in the ImageNet competition. The architecture used earlier ideas, but the result showed how much data and computing power could change their practical value.

## 2012–2017: deep learning spreads

Convolutional networks became common for image tasks. VGG explored deeper networks. Residual networks introduced skip connections that made much deeper models easier to train. Generative adversarial networks introduced a way to train generators by comparing their outputs with real examples.

Deep reinforcement learning also drew attention when systems learned to play games from images and actions.

## 2017 onward: attention and large models

The 2017 paper *Attention Is All You Need* introduced the Transformer architecture for sequence modeling. Transformers process sequence positions in parallel during training, which made large-scale training more practical.

The following years brought large language models, pretrained vision transformers, diffusion-based image generation, protein-structure prediction systems, and multimodal models. Instruction tuning and preference training made some language models easier to use as assistants.

The field continues to change. Newer systems also use tools, retrieval, and additional computation at inference time.

## A pattern to remember

Many important results combined an existing idea with something that made it practical: more data, faster hardware, a better training method, or a clearer way to evaluate it.

This does not mean every proposed method will become useful if we wait long enough. It does mean that “impossible” and “too expensive with current resources” are different claims. When you encounter a new method, ask what problem it solves and what changed to make it work.
